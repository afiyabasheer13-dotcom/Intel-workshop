```python
for column in df_numeric_columns:
  skew_value = df[column].skew()

  if -0.5 <= skew_value <= 0.5:
    print(column, "-> Use mean")
  else:
    print(column, "-> Use median")
```

```python
df["age"] = df["age"].fillna(
    df["age"].mean()
)
df["bmi"] = df["bmi"].fillna(
    df["bmi"].mean()
)
df["children"] = df["children"].fillna(
    df["children"].median()
)
```

```python
df["sex"] = df["sex"].fillna(
    df["sex"].mode()[0]
)
df["smoker"] = df["smoker"].fillna(
    df["smoker"].mode()[0]
)
df["region"] = df["region"].fillna(
    df["region"].mode()[0]
)
```

```python
import pandas as pd

df_encoded = pd.get_dummies(
    df,
    columns=["sex", "smoker", "region"],
)
df_encoded.head()
```

```python
x_df = df_encoded.drop("charges", axis=1)
y_df = df_encoded["charges"]

print("Input shape:", x_df.shape)
print("Target shape:", y_df.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Split x_df and y_df into training and testing sets first
X_train_df, X_test_df, y_train_df, y_test_df = train_test_split(
    x_df, y_df, test_size=0.2, random_state=42
)

# 2. Scale the numeric columns using insurance_scaler
insurance_scaler = StandardScaler()
scale_columns = ["age", "bmi", "children"]

X_train_df_scaled = X_train_df.copy()
X_test_df_scaled = X_test_df.copy()

X_train_df_scaled[scale_columns] = insurance_scaler.fit_transform(
    X_train_df[scale_columns]
)
X_test_df_scaled[scale_columns] = insurance_scaler.transform(
    X_test_df[scale_columns]
)
```

```python
X_train_df_scaled
```

```python
from sklearn.linear_model import LinearRegression
linear_model = LinearRegression
```

```python
from sklearn.linear_model import LinearRegression


linear_model = LinearRegression()
```

```python
# Remove the row with NaN in y_train_df (and y_test_df if present)
train_valid = y_train_df.notna()
X_train_df_scaled = X_train_df_scaled[train_valid]
y_train_df = y_train_df[train_valid]

test_valid = y_test_df.notna()
X_test_df_scaled = X_test_df_scaled[test_valid]
y_test_df = y_test_df[test_valid]

# Now train the model
linear_model.fit(X_train_df_scaled, y_train_df)
print("Linear Regression model trained successfully.")
```

```python
pred = linear_model.predict(X_test_df_scaled)
pred
```

```python
df_result = pd.DataFrame({
    "Actual Charges": y_test_df.values,
    "Predicted Charges": pred
})
df_result.head(3)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

mae = mean_absolute_error(
    y_test_df,
    pred
)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test_df, pred)
mse = mean_squared_error(y_test_df, pred)
rmse = mse ** 0.5
r2 = r2_score(y_test_df, pred)

print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
print("R2 Score:", r2)
```
